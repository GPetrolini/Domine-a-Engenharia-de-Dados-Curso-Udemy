```python
import pandas as pd
from google.cloud import bigquery
from google.oauth2 import service_account
from pandas_gbq import to_gbq
```

```python
produto   = pd.read_excel(r"C:\curso-eng-dados\Python-SGBDS\Origem de Dados\arquivos_excel\Produto.xlsx")
items     = pd.read_excel(r'C:\curso-eng-dados\Python-SGBDS\Origem de Dados/arquivos_excel/items.xlsx')
ordens    = pd.read_excel(r'C:\curso-eng-dados\Python-SGBDS\Origem de Dados\arquivos_excel\Ordens.xlsx')
categoria = pd.read_excel(r'C:\curso-eng-dados\Python-SGBDS\Origem de Dados\arquivos_excel\Categoria.xlsx')
clientes = pd.read_csv  (r'C:\curso-eng-dados\Python-SGBDS\Origem de Dados\arquivos_csv\Clientes.csv',delimiter=',')
```

```python
projeto     = 'pythoncurso-462414'
dataset     = 'Ecommerce'
BQproduto   = 'Produto'
BQitems     = 'Items'
BQordens    = 'Ordens'
BQCategoria = 'Categoria'
BQclientes  = 'clientes'
parametro   = 'replace' 
```

```python

credencial = service_account.Credentials.from_service_account_file(
 r'C:\curso-eng-dados\Python-SGBDS\Origem de Dados\pythoncurso-462414-f5463d5035ab.json', 
  scopes=['https://www.googleapis.com/auth/bigquery']
)
```

```python
produto.to_gbq(destination_table=f'{projeto}.{dataset}.{BQproduto}',
  project_id=projeto,
  if_exists=parametro,
  credentials=credencial)
```

```python
items.to_gbq(destination_table=f'{projeto}.{dataset}.{BQitems}',   
          project_id=projeto,
          if_exists=parametro,
          credentials=credencial)
```

```python
ordens.to_gbq(destination_table=f'{projeto}.{dataset}.{BQordens}',   
          project_id=projeto,
          if_exists=parametro,
          credentials=credencial)
```

```python
categoria.to_gbq(destination_table=f'{projeto}.{dataset}.{BQCategoria}',   
          project_id=projeto,
          if_exists=parametro,
          credentials=credencial)
```

```python
# Realiza a carga de tabela de clientes 
clientes.to_gbq(destination_table=f'{projeto}.{dataset}.{BQclientes}',   
          project_id=projeto,
          if_exists=parametro,
          credentials=credencial)
```
